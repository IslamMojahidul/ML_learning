2. Given a string,print the frequency of the words

```python
string = "apple mango apple orange orange apple guava mango mango"

words = string.split()

count = {}

for word in words:

    count[word] = count.get(word,0) +1
for k,v in count.items():
    print(k,v)
    
print(count)

```
