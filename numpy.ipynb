## Array Reshaping ,Ravel,flatten

```python
# Reshape
arr = np.array([1,2,3,4,5,6])
reshaped = arr.reshape(2,3)
print(reshaped)
```

```python
reshaped2 = reshaped.reshape(3,2)
print(reshaped2)
```

```python
# Ravel -> convert 1D array
ravel = reshaped2.ravel()
ravel
```

```python
ravel[0] = 100
print(ravel)
print(reshaped2)
```

```python
# Flatten -> to convert 1D array
# returns a copy of the array
flat = reshaped2.flatten()
flat
```

```python
flat[0] = 1
flat
```

```python
reshaped2
```
