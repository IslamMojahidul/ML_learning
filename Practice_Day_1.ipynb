## Smart Light

```python
n = input()

num =n.split()
brightness = float(num[0])
thresold = float(num[1])

if brightness >= thresold :
    print("ON")
else:
    print("OFF")
```

### Sensor Calibrator — Min-Max Scaling

```python
x, min_v,max_v =map(float ,input().split())
#print(x,min_v,max_v)

norm = (x - min_v) / (max_v - min_v)

print(f"{norm:.2}")
```

## Mini Loss Tracker

```python
n = int(input())
target = float(input())

#loss = list(map(float,input().split()))

#print(*loss)

suma = 0

for i in range(n):
    loss = float(input())
    suma += loss
avg = suma/n
if avg <= target:
    print("PASS")
else:
    print("RETRY") 
```

### Majority Vote Gate

```python
n = int(input())
yes_count =0
no_count =0

for i in range(n):
    
    answer = input().upper()
    
    if answer == "YES" :
        yes_count += 1
    elif answer == "NO" :
        no_count +=1
if yes_count >= no_count:
    print("ACCEPT")
else:
    print("REJECT")

```
